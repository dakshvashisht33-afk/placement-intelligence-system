````python
# ============================================================
# HELPER FUNCTIONS
# ============================================================

def convert_to_set(value):
    if pd.isna(value):
        return set()
    return {
        item.strip().lower()
        for item in str(value).split(",")
        if item.strip()
    }

def clean_text_list(value):
    return sorted([
        item.strip()
        for item in str(value).split(",")
        if item.strip()
    ])

def calculate_match(student_value, company_value):
    student_set = convert_to_set(student_value)
    company_set = convert_to_set(company_value)

    if not company_set:
        return 0.0

    return round(
        len(student_set.intersection(company_set))
        / len(company_set) * 100,
        2
    )

def get_missing_items(student_value, company_value):
    student_set = convert_to_set(student_value)
    company_set = convert_to_set(company_value)
    return sorted(company_set - student_set)

print("Helper functions loaded.")
````

````python
# ============================================================
# INPUT A NEW STUDENT
# ============================================================

def auto_fill_student_scores(student):
    """Generate aptitude and technical scores from the profile.
    The user does NOT enter these scores manually."""
    aptitude = _auto_aptitude(
        float(student["cgpa"]),
        float(student["tenth"]),
        float(student["twelfth"]),
        student["soft_skills"],
        int(student["projects"]),
        int(student["internships"])
    )
    technical = _auto_technical(
        float(student["cgpa"]),
        float(student["tenth"]),
        float(student["twelfth"]),
        student["skills"],
        student["project_type"],
        int(student["internships"])
    )
    student["aptitude_score"] = aptitude
    student["technical_score"] = technical
    return student

def input_student():
    print("\n" + "=" * 65)
    print("ENTER STUDENT INFORMATION")
    print("=" * 65)

    student = {
        "student_id": input("Student ID: ").strip(),
        "name": input("Student Name: ").strip(),
        "department": input("Department/Branch: ").strip(),
        "cgpa": float(input("CGPA: ")),
        "tenth": float(input("10th Percentage: ")),
        "twelfth": float(input("12th Percentage: ")),
        "backlogs": int(input("Number of Backlogs: ")),
        "skills": input("Technical Skills (comma separated): ").strip(),
        "soft_skills": input("Soft Skills (comma separated): ").strip(),
        "projects": int(input("Number of Projects: ")),
        "project_name": input("Main/Specialized Project Name: ").strip(),
        "project_type": input(
            "Project Type (e.g. Data Analytics / ML / Cloud / Software / Business Analytics): "
        ).strip(),
        "internships": int(input("Number of Internships: ")),
        "internship_type": input(
            "Internship Type (e.g. Data Analytics / ML / Cloud / Software / Business Analytics / None): "
        ).strip(),
        "target_role": input(
            "Target Role (Data Analyst / ML Engineer / Cloud Engineer / SDE / Business Analyst): "
        ).strip(),
        "placed": 0,
        "salary": 0
    }

    # Aptitude and technical scores are deliberately NOT requested.
    student = auto_fill_student_scores(student)

    print(f"\nAuto-filled Aptitude Score: {student['aptitude_score']}")
    print(f"Auto-filled Technical Score: {student['technical_score']}")

    return pd.Series(student)
````

````python
# ============================================================
# ACADEMIC ELIGIBILITY
# ============================================================

def check_academic_eligibility(student, company):
    reasons = []

    if float(student["cgpa"]) < float(company["min_cgpa"]):
        reasons.append(
            f"CGPA below required {company['min_cgpa']}"
        )

    if int(student["backlogs"]) > int(company["max_backlogs"]):
        reasons.append(
            f"Backlogs exceed allowed maximum of {company['max_backlogs']}"
        )

    return len(reasons) == 0, reasons
````

````python
# ============================================================
# DETERMINISTIC PLACEMENT SCORING
# ============================================================
# Python handles objective calculations. Gemini handles interpretation.
# This separation makes the system faster, cheaper, and easier to audit.

ROLE_LIST = [
    "Data Analyst",
    "ML Engineer",
    "Cloud Engineer",
    "SDE",
    "Business Analyst"
]

def calculate_numeric_score(value, minimum=0, maximum=100):
    return float(np.clip(float(value), minimum, maximum))

def analyze_student_company(student, company):
    eligible, reasons = check_academic_eligibility(student, company)

    skill_match = calculate_match(student["skills"], company["required_skills"])
    soft_score = calculate_match(student["soft_skills"], company["required_soft_skills"])
    aptitude_score = calculate_numeric_score(student["aptitude_score"])
    technical_assessment = calculate_numeric_score(student["technical_score"])

    student_role = str(student.get("target_role", "")).strip().lower()
    company_role = str(company["role"]).strip().lower()
    role_score = 100 if student_role == company_role else 0

    min_cgpa = float(company["min_cgpa"])
    student_cgpa = float(student["cgpa"])
    academic_score = min(student_cgpa / min_cgpa * 100, 100) if min_cgpa > 0 else 100

    internships = int(student.get("internships", 0))
    projects = int(student.get("projects", 0))
    project_type = str(student.get("project_type", "")).lower()
    internship_type = str(student.get("internship_type", "")).lower()
    company_role_key = str(company["role"]).lower()

    role_terms = {
        "data analyst": ["data", "analytics", "business intelligence", "financial", "customer"],
        "ml engineer": ["machine learning", "deep learning", "nlp", "predictive", "risk"],
        "cloud engineer": ["cloud", "devops", "infrastructure", "security", "reliability"],
        "sde": ["software", "backend", "full stack", "web", "mobile"],
        "business analyst": ["business", "product", "market", "operations", "fintech"]
    }
    terms = role_terms.get(company_role_key, [])

    project_specialization = 100 if any(t in project_type for t in terms) else 45
    internship_specialization = 100 if any(t in internship_type for t in terms) else (35 if internships == 0 else 55)

    experience_score = min(
        0.30 * min(projects / 3 * 100, 100) +
        0.25 * min(internships / 2 * 100, 100) +
        0.25 * project_specialization +
        0.20 * internship_specialization,
        100
    )

    fit_score = (
        academic_score * 0.15 +
        skill_match * 0.15 +
        technical_assessment * 0.18 +
        aptitude_score * 0.12 +
        soft_score * 0.12 +
        role_score * 0.10 +
        experience_score * 0.18
    )

    missing_technical = get_missing_items(student["skills"], company["required_skills"])
    missing_soft = get_missing_items(student["soft_skills"], company["required_soft_skills"])

    return {
        "company": company["company"],
        "industry": company["industry"],
        "role": company["role"],
        "package": company.get("package", 0),
        "eligible": bool(eligible),
        "academic_score": round(academic_score, 2),
        "technical_skill_score": round(skill_match, 2),
        "technical_assessment_score": round(technical_assessment, 2),
        "aptitude_score": round(aptitude_score, 2),
        "soft_skill_score": round(soft_score, 2),
        "role_score": role_score,
        "experience_score": round(experience_score, 2),
        "project_specialization_score": round(project_specialization, 2),
        "internship_specialization_score": round(internship_specialization, 2),
        "fit_score": round(fit_score, 2),
        "missing_technical_skills": missing_technical,
        "missing_soft_skills": missing_soft,
        "eligibility_reasons": reasons
    }

def compute_all_company_results(student):
    rows = [analyze_student_company(student, company) for _, company in companies_df.iterrows()]
    return pd.DataFrame(rows).sort_values("fit_score", ascending=False).reset_index(drop=True)
````

````python
# ============================================================
# COMPACT DATA PREPARATION FOR GEMINI
# ============================================================

def prepare_student_profile(student):
    return {
        "student_id": str(student.get("student_id", "")),
        "name": str(student.get("name", "")),
        "department": str(student.get("department", "")),
        "target_role": str(student.get("target_role", "")),
        "academics": {
            "cgpa": float(student.get("cgpa", 0)),
            "10th_percentage": float(student.get("tenth", 0)),
            "12th_percentage": float(student.get("twelfth", 0)),
            "backlogs": int(student.get("backlogs", 0))
        },
        "technical_skills": clean_text_list(student.get("skills", "")),
        "soft_skills": clean_text_list(student.get("soft_skills", "")),
        "projects": {
            "count": int(student.get("projects", 0)),
            "main_project": str(student.get("project_name", "")),
            "type": str(student.get("project_type", ""))
        },
        "internships": {
            "count": int(student.get("internships", 0)),
            "type": str(student.get("internship_type", ""))
        },
        "assessments": {
            "aptitude_score": float(student.get("aptitude_score", 0)),
            "technical_score": float(student.get("technical_score", 0))
        }
    }

def prepare_company_profile(company):
    return {
        "company": str(company["company"]),
        "industry": str(company["industry"]),
        "role": str(company["role"]),
        "minimum_cgpa": float(company["min_cgpa"]),
        "maximum_backlogs": int(company["max_backlogs"]),
        "required_technical_skills": clean_text_list(company["required_skills"]),
        "required_soft_skills": clean_text_list(company["required_soft_skills"]),
        "package": int(company.get("package", 0)),
        "visit_month": str(company.get("visit_month", ""))
    }

def compact_result(row):
    return {
        "company": row["company"],
        "industry": row["industry"],
        "role": row["role"],
        "package": int(row.get("package", 0)),
        "eligible": bool(row["eligible"]),
        "fit_score": float(row["fit_score"]),
        "academic_score": float(row["academic_score"]),
        "technical_skill_score": float(row["technical_skill_score"]),
        "technical_assessment_score": float(row["technical_assessment_score"]),
        "aptitude_score": float(row["aptitude_score"]),
        "soft_skill_score": float(row["soft_skill_score"]),
        "experience_score": float(row["experience_score"]),
        "missing_technical_skills": row["missing_technical_skills"],
        "missing_soft_skills": row["missing_soft_skills"],
        "eligibility_reasons": row["eligibility_reasons"]
    }
````

````python
# ============================================================
# GEMINI PROMPT + STRUCTURED JSON RESPONSE HANDLING
# ============================================================

GEMINI_SYSTEM_PROMPT = """
You are the Placement Intelligence Engine for a university placement system.

Analyze ONLY the student and company information supplied in the request.
Python has already calculated objective eligibility and profile-fit metrics.
Your job is to interpret those metrics and produce actionable placement
intelligence.

Rules:
- Never invent qualifications, skills, companies, requirements, scores, or outcomes.
- Never claim that selection is guaranteed.
- fit_score is an analytical profile-fit indicator, NOT a real probability of selection.
- Eligibility and fit are different: a student may be eligible but have skill gaps.
- Prioritize the highest-impact gaps.
- Consider technical skills, soft skills, academics, projects, internships,
  aptitude, technical assessment, target role, and company requirements.
- Give concrete actions that a BTech student can complete.
- Do not treat synthetic data as real-world placement evidence.
- If information is missing, say that it is unavailable rather than guessing.
- Keep the analysis concise but useful.
"""

# Gemini structured output schema.
# Using JSON schema means the notebook does not depend on fragile
# markdown/JSON parsing from free-form model responses.
GEMINI_JSON_SCHEMA = {
    "type": "object",
    "properties": {
        "overall_assessment": {"type": "string"},
        "target_role_analysis": {
            "type": "object",
            "properties": {
                "role": {"type": "string"},
                "fit_level": {"type": "string"},
                "reasoning": {"type": "string"}
            },
            "required": ["role", "fit_level", "reasoning"]
        },
        "strengths": {
            "type": "array",
            "items": {"type": "string"}
        },
        "critical_gaps": {
            "type": "array",
            "items": {"type": "string"}
        },
        "technical_gaps": {
            "type": "array",
            "items": {"type": "string"}
        },
        "soft_skill_gaps": {
            "type": "array",
            "items": {"type": "string"}
        },
        "learning_priorities": {
            "type": "array",
            "items": {
                "type": "object",
                "properties": {
                    "priority": {"type": "integer"},
                    "skill": {"type": "string"},
                    "reason": {"type": "string"},
                    "action": {"type": "string"}
                },
                "required": [
                    "priority",
                    "skill",
                    "reason",
                    "action"
                ]
            }
        },
        "project_recommendations": {
            "type": "array",
            "items": {"type": "string"}
        },
        "internship_recommendations": {
            "type": "array",
            "items": {"type": "string"}
        },
        "company_analysis": {
            "type": "array",
            "items": {
                "type": "object",
                "properties": {
                    "company": {"type": "string"},
                    "role": {"type": "string"},
                    "eligibility": {"type": "string"},
                    "fit_summary": {"type": "string"},
                    "matching_skills": {
                        "type": "array",
                        "items": {"type": "string"}
                    },
                    "missing_skills": {
                        "type": "array",
                        "items": {"type": "string"}
                    },
                    "recommended_actions": {
                        "type": "array",
                        "items": {"type": "string"}
                    }
                },
                "required": [
                    "company",
                    "role",
                    "eligibility",
                    "fit_summary",
                    "matching_skills",
                    "missing_skills",
                    "recommended_actions"
                ]
            }
        },
        "career_roadmap": {
            "type": "array",
            "items": {
                "type": "object",
                "properties": {
                    "phase": {"type": "string"},
                    "duration": {"type": "string"},
                    "objectives": {
                        "type": "array",
                        "items": {"type": "string"}
                    },
                    "deliverables": {
                        "type": "array",
                        "items": {"type": "string"}
                    }
                },
                "required": [
                    "phase",
                    "duration",
                    "objectives",
                    "deliverables"
                ]
            }
        }
    },
    "required": [
        "overall_assessment",
        "target_role_analysis",
        "strengths",
        "critical_gaps",
        "technical_gaps",
        "soft_skill_gaps",
        "learning_priorities",
        "project_recommendations",
        "internship_recommendations",
        "company_analysis",
        "career_roadmap"
    ]
}


def _extract_json(text):
    text=(text or '').strip()
    text=re.sub(r'^```(?:json)?\s*','',text,flags=re.I)
    text=re.sub(r'\s*```$','',text)
    try: return json.loads(text)
    except json.JSONDecodeError:
        match=re.search(r'\{.*\}',text,flags=re.S)
        if match: return json.loads(match.group(0))
        raise

def _is_retryable_error(exc):
    t=str(exc).lower()
    return any(x in t for x in ['429','resource_exhausted','rate limit','500','502','503','504','unavailable','internal','high demand','temporarily','deadline exceeded'])

def _gemini_config():
    return types.GenerateContentConfig(system_instruction=GEMINI_SYSTEM_PROMPT,response_mime_type='application/json',response_schema=GEMINI_JSON_SCHEMA,temperature=0.2)

def _generate_gemini(contents,stream=False,max_attempts=3):
    errors=[]
    for model_name in GEMINI_MODEL_CANDIDATES:
        for attempt in range(1,max_attempts+1):
            try:
                config=_gemini_config()
                if stream:
                    return model_name,gemini_client.models.generate_content_stream(model=model_name,contents=contents,config=config)
                return model_name,gemini_client.models.generate_content(model=model_name,contents=contents,config=config)
            except Exception as exc:
                errors.append(f'{model_name} attempt {attempt}: {type(exc).__name__}: {exc}')
                if not _is_retryable_error(exc) or attempt>=max_attempts: break
                delay=min(2**(attempt-1),8); print(f'Gemini {model_name} temporary error; retrying in {delay}s...'); time.sleep(delay)
    raise RuntimeError('Gemini API request failed across available Gemini 3.x models.\n'+'\n'.join(errors[-12:]))

def _response_text(response): return getattr(response,'text',None) or ''

def gemini_analysis(student,result_df,analysis_type='general',selected_role=None,show_stream=False):
    if analysis_type=='specific':
        candidates=result_df[result_df['role'].astype(str).str.lower()==selected_role.lower()].copy().sort_values('fit_score',ascending=False).head(12)
    else:
        candidates=pd.concat([result_df.head(8),result_df[result_df['eligible']].head(8),result_df.groupby('role',group_keys=False).head(2)]).drop_duplicates('company').sort_values('fit_score',ascending=False).head(15)
    companies_for_ai=[]
    for _,row in candidates.iterrows():
        cn=str(row['company']).strip().lower(); rn=str(row['role']).strip().lower()
        matches=companies_df[(companies_df['company'].astype(str).str.strip().str.lower()==cn)&(companies_df['role'].astype(str).str.strip().str.lower()==rn)]
        if not matches.empty: companies_for_ai.append(prepare_company_profile(matches.iloc[0]))
    payload={'analysis_type':analysis_type,'selected_role':selected_role,'student':prepare_student_profile(student),'deterministic_metrics':[compact_result(row) for _,row in candidates.iterrows()],'companies':companies_for_ai}
    user_prompt = (
        'Analyze this university placement profile.\n\n'
        'Return JSON matching the supplied response schema.\n\n'
        'DATA:\n'+json.dumps(payload,indent=2,default=str)+'\n\n'
        'Instructions:\n'
        '- learning_priorities: at most 8 items.\n'
        '- company_analysis: analyze every supplied company.\n'
        '- career_roadmap: 3 to 5 practical phases.\n'
        '- Base every statement on supplied data.\n'
        '- Do not convert fit_score into a selection probability.\n'
        '- Do not invent tools, qualifications, skills, companies, requirements, scores, or outcomes.'
    )
    if not show_stream:
        model_name,response=_generate_gemini(user_prompt,stream=False); text=_response_text(response)
        if not text.strip(): raise RuntimeError(f'Gemini {model_name} returned an empty response.')
        return _extract_json(text),candidates
    print('\n'+'='*20+' Gemini Response Stream '+'='*20); chunks=[]; stream_error=None
    try:
        model_name,event_stream=_generate_gemini(user_prompt,stream=True)
        for chunk in event_stream:
            t=getattr(chunk,'text',None)
            if t: chunks.append(t); print(t,end='',flush=True)
        print(); text=''.join(chunks)
        if text.strip(): return _extract_json(text),candidates
        raise RuntimeError('Gemini streaming returned no text.')
    except Exception as exc:
        stream_error=exc; print('\n[Streaming failed. Switching to standard Gemini generation...]'); print(f'Streaming error: {type(exc).__name__}: {exc}')
    try:
        model_name,response=_generate_gemini(user_prompt,stream=False); text=_response_text(response)
        if not text.strip(): raise RuntimeError(f'Gemini {model_name} returned an empty recovery response.')
        print(f'Standard Gemini recovery successful using {model_name}.')
        return _extract_json(text),candidates
    except Exception as recovery_error:
        raise RuntimeError('Gemini analysis failed after streaming and standard recovery.\n'+f'Streaming error: {stream_error}\nRecovery error: {recovery_error}') from recovery_error
````

````python
# GEMINI CONNECTION TEST
def test_gemini_connection():
    print("\n"+"="*70); print("GEMINI CONNECTION TEST"); print("="*70)
    try:
        model_name,response=_generate_gemini("Return JSON with exactly one field named status. Set it to READY.",stream=False,max_attempts=2)
        result=_extract_json(_response_text(response)); print("Connection: ✓"); print("Structured JSON: ✓"); print("Working model:",model_name); print("Response:",result); return True
    except Exception as exc:
        print("Connection: ✗"); print(type(exc).__name__,exc); return False
````

````python
# ============================================================
# AI REPORT DISPLAY
# ============================================================

def display_ai_report(ai):
    print("\n" + "=" * 90)
    print("PLACEMENT INTELLIGENCE REPORT")
    print("=" * 90)

    print("\nOVERALL ASSESSMENT")
    print("-" * 90)
    print(ai.get("overall_assessment", "Not available"))

    role = ai.get("target_role_analysis", {})
    print("\nTARGET ROLE")
    print("-" * 90)
    print("Role:", role.get("role", ""))
    print("Fit level:", role.get("fit_level", ""))
    print("Reasoning:", role.get("reasoning", ""))

    sections = [
        ("STRENGTHS", "strengths"),
        ("CRITICAL GAPS", "critical_gaps"),
        ("TECHNICAL GAPS", "technical_gaps"),
        ("SOFT-SKILL GAPS", "soft_skill_gaps"),
        ("PROJECT RECOMMENDATIONS", "project_recommendations"),
        ("INTERNSHIP RECOMMENDATIONS", "internship_recommendations")
    ]
    for title, key in sections:
        print(f"\n{title}\n" + "-" * 90)
        items = ai.get(key, [])
        if not items:
            print("None identified from the supplied data.")
        for item in items:
            print("•", item)

    print("\nLEARNING PRIORITIES")
    print("-" * 90)
    for item in ai.get("learning_priorities", []):
        print(f"{item.get('priority', '')}. {item.get('skill', '')}")
        print("   Why:", item.get("reason", ""))
        print("   Action:", item.get("action", ""))

    print("\nCOMPANY ANALYSIS")
    print("-" * 90)
    for company in ai.get("company_analysis", []):
        print(f"\n{company.get('company', '')} — {company.get('role', '')}")
        print("Eligibility:", company.get("eligibility", ""))
        print("Fit:", company.get("fit_summary", ""))
        print("Matching skills:", ", ".join(company.get("matching_skills", [])))
        print("Missing skills:", ", ".join(company.get("missing_skills", [])))
        for action in company.get("recommended_actions", []):
            print("  •", action)

    print("\nCAREER ROADMAP")
    print("-" * 90)
    for phase in ai.get("career_roadmap", []):
        print(f"\n{phase.get('phase', '')} ({phase.get('duration', '')})")
        for objective in phase.get("objectives", []):
            print("  Objective:", objective)
        for deliverable in phase.get("deliverables", []):
            print("  Deliverable:", deliverable)
````

````python
# ============================================================
# MAIN INTERACTIVE SYSTEM
# ============================================================

def run_placement_analysis():
    print("\n" + "=" * 90)
    print("UNIVERSITY PLACEMENT INTELLIGENCE SYSTEM — GOOGLE GEMINI")
    print("=" * 90)

    use_new_student = input("\nDo you want to enter a new student? (yes/no): ").strip().lower()

    if use_new_student in ["yes", "y"]:
        student = add_new_student()
    else:
        student_id = input("Enter existing Student ID (e.g. S001): ").strip()
        matches = students_df[students_df["student_id"].astype(str).str.lower() == student_id.lower()]
        if matches.empty:
            print("Student ID not found.")
            return
        student = matches.iloc[0]

    print(f"\nStudent: {student['name']} | Target role: {student['target_role']}")
    print(f"Companies in reference database: {len(companies_df)}")
    print(f"Auto aptitude score: {student['aptitude_score']}")
    print(f"Auto technical score: {student['technical_score']}")

    print("\nChoose analysis:")
    print("1. General AI placement intelligence")
    print("2. Specific-role AI placement intelligence")
    print("3. General AI placement intelligence + streamed Gemini response")

    choice = input("Enter 1, 2, or 3: ").strip()

    all_results = compute_all_company_results(student)

    if choice == "1":
        ai, candidates = gemini_analysis(student, all_results, "general", show_stream=False)
    elif choice == "3":
        ai, candidates = gemini_analysis(student, all_results, "general", show_stream=True)
    elif choice == "2":
        print("\nAvailable roles:")
        for i, role in enumerate(ROLE_LIST, 1):
            print(f"{i}. {role}")
        role_input = input("\nEnter role name or number: ").strip()
        if role_input.isdigit() and 1 <= int(role_input) <= len(ROLE_LIST):
            selected_role = ROLE_LIST[int(role_input) - 1]
        else:
            selected_role = role_input
        if selected_role.lower() not in [r.lower() for r in ROLE_LIST]:
            print("Invalid role. Choose one of:", ", ".join(ROLE_LIST))
            return
        ai, candidates = gemini_analysis(
            student, all_results, "specific", selected_role=selected_role, show_stream=False
        )
    else:
        print("Invalid choice.")
        return

    print("\nCompanies supplied to Gemini:", len(candidates))
    display_ai_report(ai)

    print("\n" + "=" * 90)
    print("DETERMINISTIC TOP COMPANY FITS")
    print("=" * 90)
    display(
        all_results[[
            "company", "role", "fit_score", "eligible",
            "technical_skill_score", "soft_skill_score", "experience_score"
        ]].head(15)
    )

    return {
        "student": student,
        "deterministic_results": all_results,
        "gemini_analysis": ai,
        "gemini_companies": candidates
    }

print("\nSystem ready. Run: run_placement_analysis()")
````

````python
# ============================================================
# OPTIONAL PROGRAMMATIC API
# ============================================================

def analyze_existing_student(student_id, analysis_type="general", role=None, show_stream=False):
    matches = students_df[
        students_df["student_id"].astype(str).str.lower() == str(student_id).lower()
    ]
    if matches.empty:
        raise ValueError(f"Student ID not found: {student_id}")

    student = matches.iloc[0]
    all_results = compute_all_company_results(student)

    if analysis_type.lower() == "general":
        ai, candidates = gemini_analysis(
            student, all_results, "general", show_stream=show_stream
        )
    elif analysis_type.lower() == "specific":
        if not role:
            raise ValueError("role is required for specific analysis")
        ai, candidates = gemini_analysis(
            student, all_results, "specific", selected_role=role, show_stream=show_stream
        )
    else:
        raise ValueError("analysis_type must be 'general' or 'specific'")

    display_ai_report(ai)
    return {
        "student": student,
        "deterministic_results": all_results,
        "gemini_analysis": ai,
        "gemini_companies": candidates
    }

# Examples:
# result = analyze_existing_student("S001", "general")
# result = analyze_existing_student("S001", "specific", "Cloud Engineer")
````
